# IPL Matches(2008-2024) Exploratory Data Analysis

### Importing Important Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Loading the Dataset:

```python
df = pd.read_csv('matches.csv')
```

```python
df.head()
```

```python
df.columns
```

### Matches Dataset Contains the following Columns:
- <b>id:</b> A unique number identifying the specific match.
- <b>season:</b> The year or season of the IPL.
- <b>date:</b> The specific date when the match was played.
- <b>city:</b> The city where the match took place.
- <b>venue:</b> The specific stadium name.
- <b>team1 & team2:</b> The names of the two teams playing the match.
- <b>toss_winner:</b> The team that won the coin toss.
- <b>toss_decision:</b> The decision made by the toss winner (either bat or field).
- <b>winner:</b> The name of the team that won the match.
- <b>result:</b> The method of victory.
- <b>result_margin:</b> The numerical margin of victory.
- <b>player_of_match:</b> The player awarded "Man of the Match."
- <b>match_type:</b> The stage of the tournament (e.g., League, Qualifier, Final).
- <b>target_runs:</b> The score the chasing team needed to win.
- <b>target_overs:</b> The number of overs allotted.
- <b>super_over:</b> Y or N indicating if the match went to a Super Over tie-breaker.
- <b>method:</b> Usually indicates if the D/L method (Duckworth-Lewis) was used due to rain. NA means it was a normal match.
- <b>umpire1 & umpire2:</b> The names of the on-field umpires officiating the match.

### Now Let's Start Some Exploratory Data Analysis 

```python
# Fix Team Names
team_mapping = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Deccan Chargers': 'Sunrisers Hyderabad',
    'Rising Pune Supergiants': 'Rising Pune Supergiant',
    'Pune Warriors': 'Rising Pune Supergiant', 
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru'
}
df.replace(team_mapping, inplace=True)
```

Mapping the teams previous names with their current ones for better analysis

```python
plt.figure(figsize=(12, 6))
sns.countplot(x='season', data=df,hue='season',legend=False, order=df['season'].value_counts().index.sort_values(), palette='cool')
plt.xlabel('Season')
plt.ylabel('No. of matches')
plt.title('No. of matches played on each IPL Season')
```

#### The above graph shows the total no. of matches played in each IPL Seasons.
<b>Observation:</b> Most matches were played in IPL Season 2013

```python
df_player_of_match = df['player_of_match'].value_counts().reset_index()[:10]
```

```python
df_player_of_match
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(x='player_of_match', y='count', data=df_player_of_match, hue='player_of_match', palette='inferno')
plt.title('Top 10 players winning the player of the match award')
```

#### The above graph shows the most player of the match won by the players.
<b>Observation:</b> AB de Villiers won the most player of the match award followed by CH Gayle and RG Sharma

```python
df_city = df['city'].value_counts().reset_index()[:10]
```

```python
df_city
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(x='city', y='count', data=df_city, hue='city', palette='inferno')
plt.title('Total No. Of Matches PLeyed in Each Cities')
plt.ylabel('Matches Count')
```

<b>Observation:</b> Most matches are played in Mumbai

```python
df_matches_wins = df['winner'].value_counts().reset_index()
```

```python
df_matches_wins
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(x='count', y='winner', data=df_matches_wins, hue='winner', palette='inferno')
plt.title('Total matches won by each teams')
plt.xlabel('Total No. of Wins')
plt.ylabel('Teams')
plt.tight_layout()
```

<b>Observation:</b> Most matches are won by Mumbai Indians followed by Chennai Super Kings and Kolkata Knight Riders

```python
matches_played = df['team1'].value_counts() + df['team2'].value_counts()
```

```python
matches_won = df['winner'].value_counts()
```

```python
win_percentage = (matches_won/matches_played * 100).sort_values(ascending=False)[:5]
```

```python
win_percentage
```

```python
plt.pie(x=win_percentage.values, labels=win_percentage.index, autopct='%1.1f%%', startangle=90)
plt.title('Top 5 teams with highest win percentage')
```

<b>Observation:</b> Gujarat Titans have the highest win percentage

```python
df_toss_decision = df['toss_decision'].value_counts().reset_index()
```

```python
df_toss_decision
```

```python
wedgeprops = {"edgecolor" : "black",
                      'linewidth': 1,
                      'antialiased': True}
plt.pie(x='count', labels='toss_decision', data=df_toss_decision, autopct='%1.0f%%', wedgeprops=wedgeprops, colors=['#ff7f50','#87cefa'])
plt.title('Toss Decision')
```

<b>Observation:</b> Teams choose to field more than to bat by winning the toss

```python
df['toss_match_winner'] = df['toss_winner'] == df['winner']
```

```python
toss_impact = df['toss_match_winner'].value_counts(normalize=True) * 100
```

```python
plt.figure(figsize=(6, 6))
plt.pie(toss_impact, labels=['Toss Winner Won Match', 'Toss Winner Lost Match'], 
        autopct='%1.1f%%', colors=['#66b3ff', '#ff9999'], startangle=90)
plt.title("Impact of Winning the Toss on Match Result")
plt.show()
```

<b>Observation:</b> Toss decision does not make much impact on match result

```python
plt.figure(figsize=(10, 5))
sns.countplot(x='season', hue='toss_decision', data=df, palette='Set2')
plt.title('Toss Decision across Seasons')
plt.xticks(rotation=45)
plt.show()
```

<b>Observation:</b> Teams likes to choose to field more than to bat in later IPL Seasons
